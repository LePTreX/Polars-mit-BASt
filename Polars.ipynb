```python
import polars as pl

DATA         =   "2024_A_S.txt"
TABLE        =   "Jawe2024.csv"
THRESHOLD    =   0.30
WEEKTYPE     =   "w"

print("Polars", pl.__version__, "| Threads:", pl.thread_pool_size())

```

```python
raw = pl.scan_csv(DATA, separator=";", infer_schema_length=0)

print(raw.collect_schema().len(), "Spalten")
raw.head().collect()
```

```python
VALUE   =   ["KFZ_R1", "KFZ_R2", "Lkw_R1", "Lkw_R2"]
FLAGS   =   [f"K_{c}" for c in VALUE]

select = raw.select("Zst", "Land", "Datum", "Wotag", "Fahrtzw", "Stunde", *VALUE, *FLAGS)
select.head().collect()
```

```python
clean = (
    select
    .with_columns(pl.all().str.strip_chars())
    .with_columns(
        pl.col("Zst").str.zfill(4),
        pl.col("Datum").str.to_date("%y%m%d"),
        pl.col("Stunde").cast(pl.Int8),
        pl.col("Wotag").cast(pl.Int8),
    )
)
clean.head().collect()
```

```python
clean.group_by("K_KFZ_R1").len().sort("len", descending=True).collect()
```

```python
clean.filter(pl.col("K_KFZ_R1") == "s").head().collect()
```

```python
join = (
    pl.read_csv(
        TABLE, separator=";", encoding="latin1", infer_schema_length=0,
        columns=["DZ_Nr", "DZ_Name", "Land_Code", "Str_Kl", "Str_Nr", "Koor_UTM32_E", "Koor_UTM32_N"],
    )
    .rename({"DZ_Nr":"Zst"})
    .with_columns(
        pl.col("Koor_UTM32_E", "Koor_UTM32_N").str.replace_all(".","", literal=True).cast(pl.Int64)
    )
)
join.head(10)
```

```python
ganglinie = (
    join
    .filter(pl.col("Fahrtzw") == WEEKTYPE)
    .with_columns(
        (pl.col("KFZ_R1") + pl.col("KFZ_R2")).alias("Kfz"),
        (pl.col("Lkw_R1") + pl.col("Lkw_R2")).alias("Lkw"),
    )
    .join(join.lazy(), on="Zst", how="inner")
    .sort("Zst","Stunde")
    .collect()
)
print(Ganglinie.height, "Zeilen =", Ganglinie["Zst"].n_unique(), "Zählstellen x 24 Studnen")
ganglinie.head(8)
```
